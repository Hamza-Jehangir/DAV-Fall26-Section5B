```python
# 24I-6034 ______ HAMZA FAHEEM ________ DS-5B _______ DAV LAB 01

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("adult - adult.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# TASK 1
df.replace(['?', ' ?'], np.nan, inplace = True)
missingVals = df.isnull().sum()
missingVals
```

```python
# TASK 2
missingValsPerc = (missingVals / len(df)) * 100
missingValsPerc
```

```python
# TASK 3
columnsWithUnusualValz = ['workclass', 'occupation', 'native.country']

for c in columnsWithUnusualValz:
    df[c].fillna('Unknown', inplace = True)

print("Total number of missing values = ", df.isnull().sum().sum())    
```

```python
# TASK 4
exactDuplis = df.duplicated().sum()
print("Duplicated Rows: ", exactDuplis)

featuresOnly = [c for c in df.columns if c != 'income']
featuresDupli = df.duplicated(subset = featuresOnly).sum()
print("Duplicate rows after excluding the income column = ", featuresDupli)
```

```python
# TASK 5
colsToFocusOn = ['education', 'marital.status', 'native.country']
print("Getting unique values :>")
for c in colsToFocusOn:
    print(f"Unique values in {c} = ")
    print(df[c].unique())
    print("\n")

objColz = df.select_dtypes(include = ['object']).columns
for c in objColz:
    df[c] = df[c].str.strip()

print("<-- After applying str.strip() -->")
for c in colsToFocusOn:
    print(f"Unique Values in brand new {c} = ")
    print(df[c].unique()[:10])
    print("\n")
```

```python
# TASK 6 (A)
print(df.describe())
print("-----------------------------------------------------------------")
print(df.describe(include = ['object']))
```

```python
# TASK 6 (B)
ageMean = df['age'].mean()
ageMedian = df['age'].median()
hoursMean = df['hours.per.week'].mean()
hoursMedian = df['hours.per.week'].median()

print("Age Mean = ", ageMean)
print("Age Median = ", ageMedian)
print("Hours/Week Mean = ", hoursMean)
print("Hours/Week Median = ", hoursMedian)
```

```python
# TASK 7
# --- (A) ---
mostCommonOccupation = df['occupation'].value_counts().index[0]
print(f"Most Common Occupation = {mostCommonOccupation} \n")

# --- (B) ---
sexDistribution = df['sex'].value_counts(normalize = True) * 100
print(f"Sex Distribution = {sexDistribution}\n")

# --- (C) ---
incomeDistribution = df['income'].value_counts(normalize = True) * 100
print(f"Income Distribution = {incomeDistribution}")
```

```python
# TASK 8
df['education'] = df['education'].str.strip()

cnsistncy = df.groupby('education')['education.num'].unique()
print(cnsistncy)

inconsistentMapping = cnsistncy[cnsistncy.apply(len) > 1]

if len(inconsistentMapping) == 0:
    print("No inconsistency found :>")
else:
    print("Inconsistencies found!!!")
    print(inconsistentMapping)
```

```python
# TASK 9
import sqlite3

conn = sqlite3.connect(':memory:')
df.to_sql('adult_income', conn, index = False, if_exists = 'replace')

qFiltered = "SELECT * FROM adult_income WHERE age > 30"
df_FilteredBySQL = pd.read_sql_query(qFiltered, conn)

qAll = "SELECT * FROM adult_income"
df_All = pd.read_sql_query(qAll, conn)

print(f"Original CSV Data-Frame shape: {df.shape}")
print(f"Full SQL Extraction shape: {df_All.shape}")
print(f"Filtered SQL shape (age > 30): {df_FilteredBySQL.shape}")
conn.close()
```
